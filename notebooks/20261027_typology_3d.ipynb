# Manam volcano - above and below sea level

Explore Manam Island and its surrounding sea floor in three dimensions. Drag to orbit, scroll to zoom, switch to a north-up overhead view, or reveal 100 m contours. Horizontal and vertical axes use kilometres, with **no vertical exaggeration**.

The land elevations come from SRTM; offshore depths come from global ETOPO1 bathymetry in the Mapzen terrain tile. ETOPO1 has 1 arc-minute source resolution, much coarser than the SRTM land elevations. The blue overlay marks sea level (0 m); it is a visual water surface, not a tide model. This is not a nautical chart or a volcanic hazard map. The display is resampled to 601 x 601 points for a smoother surface; this does not add terrain detail beyond the source data.

## Data and reproducibility

This notebook uses a crop of the 1 arc-second HGT tile `S05E145`, distributed by [Mapzen Elevation Tiles on AWS](https://elevation-tiles-prod.s3.amazonaws.com/skadi/S05/S05E145.hgt.gz). Mapzen's terrain data combine [SRTM](https://www.usgs.gov/centers/eros/science/usgs-eros-archive-digital-elevation-shuttle-radar-topography-mission-srtm) land elevations with [ETOPO1 ocean bathymetry](https://www.ngdc.noaa.gov/mgg/global/global.html). Ocean depths are therefore resampled from ETOPO1's 1 arc-minute grid; the HGT file's finer sample spacing does not imply finer seafloor detail. The crop is bundled in `data/manam_dem.npz`, so normal runs are offline. Set `REFRESH_DATA = True` to rebuild it from the public tile.

SRTM data courtesy of the U.S. Geological Survey. Global ETOPO1 terrain data courtesy of the U.S. National Oceanic and Atmospheric Administration.

The HTML export is generated from this notebook and written to `output/20261027_typology_3d.html`. Terrain processing, sea surface, plot controls, and page layout are kept here as the source of truth.

```python
from pathlib import Path
import gzip
import hashlib
import json
import math
import sys
import urllib.request
import numpy as np
import plotly.graph_objects as go

current = Path.cwd().resolve()
project_root = next((p for p in (current, *current.parents)
                     if (p / 'site_export.py').is_file() and (p / 'notebooks').is_dir()), None)
if project_root is None:
    raise RuntimeError('Run this notebook from the repository or notebooks directory.')
sys.path.insert(0, str(project_root)) if str(project_root) not in sys.path else None
from site_export import write_plot_page

SUMMIT_LAT, SUMMIT_LON = -4.08, 145.037
SUMMIT_M = 1807
RADIUS_KM, GRID_SIZE = 8.0, 601
REFRESH_DATA = False
SOURCE_TILES = [
    ('S05E144', 144, 'https://elevation-tiles-prod.s3.amazonaws.com/skadi/S05/S05E144.hgt.gz',
     '9d125abd0e433ea63b883c1897baedc41756d41d283501311a494ce48122c087'),
    ('S05E145', 145, 'https://elevation-tiles-prod.s3.amazonaws.com/skadi/S05/S05E145.hgt.gz',
     'd9e1c626e769a6e1ecc5dad0ad133cb12995bf19c968410e0cc325730914cd58'),
]
SOURCE_SHA256 = {name: digest for name, _, _, digest in SOURCE_TILES}
data_file = project_root / 'data/manam_dem.npz'
cache_dir = project_root / '.cache/terrain'
km_lat = 111.195
km_lon = km_lat * math.cos(math.radians(SUMMIT_LAT))
bounds = (SUMMIT_LAT - RADIUS_KM / km_lat, SUMMIT_LAT + RADIUS_KM / km_lat,
          SUMMIT_LON - RADIUS_KM / km_lon, SUMMIT_LON + RADIUS_KM / km_lon)
```

```python
# Rebuild only on request; the committed crop makes normal runs fully offline.
if REFRESH_DATA:
    cache_dir.mkdir(parents=True, exist_ok=True)
    tiles, sources = [], []
    for name, tile_west, url, expected_sha256 in SOURCE_TILES:
        tile_file = cache_dir / f'{name}.hgt.gz'
        if not tile_file.exists():
            request = urllib.request.Request(url, headers={'User-Agent': 'phey.app-terrain/1.0'})
            with urllib.request.urlopen(request, timeout=120) as response:
                raw_download = response.read()
            if hashlib.sha256(raw_download).hexdigest() != expected_sha256:
                raise ValueError(f'Downloaded source checksum mismatch for {name}.')
            temporary = tile_file.with_suffix('.part')
            temporary.write_bytes(raw_download)
            temporary.replace(tile_file)
        compressed = tile_file.read_bytes()
        if hashlib.sha256(compressed).hexdigest() != expected_sha256:
            raise ValueError(f'Cached source checksum mismatch for {name}; remove the corrupt cache and retry.')
        samples = np.frombuffer(gzip.decompress(compressed), dtype='>i2')
        if samples.size != 3601 * 3601:
            raise ValueError(f'Unexpected HGT sample count for {name}: {samples.size}')
        tiles.append(samples.reshape((3601, 3601)).astype(np.int16))
        sources.append({'tile': name, 'url': url, 'sha256': expected_sha256})

    if SOURCE_TILES[1][1] != SOURCE_TILES[0][1] + 1:
        raise ValueError('The source tiles are not adjacent in longitude.')
    if not np.array_equal(tiles[0][:, -1], tiles[1][:, 0]):
        raise ValueError('The adjacent HGT tile columns do not align.')
    tile = np.concatenate((tiles[0][:, :-1], tiles[1]), axis=1)
    tile_lat = -4 - np.arange(3601, dtype=np.float64) / 3600
    tile_lon = SOURCE_TILES[0][1] + np.arange(7201, dtype=np.float64) / 3600
    rows = np.flatnonzero((tile_lat >= bounds[0]) & (tile_lat <= bounds[1]))
    cols = np.flatnonzero((tile_lon >= bounds[2]) & (tile_lon <= bounds[3]))
    if not rows.size or not cols.size:
        raise ValueError('The source tiles do not cover the Manam crop.')
    r0, r1 = max(0, rows[0]-1), min(tile_lat.size, rows[-1]+2)
    c0, c1 = max(0, cols[0]-1), min(tile_lon.size, cols[-1]+2)
    crop = tile[r0:r1, c0:c1]
    source_lat = tile_lat[r0:r1]
    source_lon = tile_lon[c0:c1]
    if np.any(crop == -32768):
        raise ValueError('The Manam crop contains void samples; refusing to invent elevations.')
    provenance = json.dumps({
        'dataset': 'Mapzen terrain tile: SRTM land elevation and ETOPO1 ocean bathymetry',
        'distribution': 'Mapzen Elevation Tiles (Skadi), public AWS objects',
        'tiles': [name for name, _, _, _ in SOURCE_TILES],
        'sources': sources, 'accessed': '2026-10-27',
        'source_sha256': SOURCE_SHA256,
        'format': 'HGT tiles sampled at 1 arc-second; WGS84, elevations in metres; ETOPO1 ocean source is 1 arc-minute',
        'attribution': 'SRTM data courtesy of the U.S. Geological Survey; global ETOPO1 terrain data courtesy of the U.S. National Oceanic and Atmospheric Administration.'})
    np.savez_compressed(data_file, elevation_m=crop, latitude=source_lat,
                        longitude=source_lon, provenance=provenance)

if not data_file.is_file():
    raise FileNotFoundError(f'Missing bundled terrain crop: {data_file}. Set REFRESH_DATA = True to build it.')
with np.load(data_file, allow_pickle=False) as data:
    crop = data['elevation_m'].astype(np.float64)
    source_lat, source_lon = data['latitude'], data['longitude']
    provenance = json.loads(str(data['provenance']))
if (provenance.get('source_sha256') != SOURCE_SHA256
        or provenance.get('tiles') != [name for name, _, _, _ in SOURCE_TILES]
        or provenance.get('dataset') != 'Mapzen terrain tile: SRTM land elevation and ETOPO1 ocean bathymetry'):
    raise ValueError('Unexpected cached dataset or source tiles; rebuild with REFRESH_DATA = True')
if crop.shape != (len(source_lat), len(source_lon)):
    raise ValueError('Terrain grid and coordinate arrays do not align.')
if np.any(crop == -32768) or not np.isfinite(crop).all():
    raise ValueError('Terrain contains void or invalid elevations.')
if not (np.all(np.diff(source_lat) < 0) and np.all(np.diff(source_lon) > 0)):
    raise ValueError('Unexpected HGT coordinate order.')
print(f'Loaded source crop: {crop.shape}; {data_file.stat().st_size / 1024:.1f} KiB')
```

```python
# Bilinear sampling uses the original one-arc-second grid without altering the source.
lat = np.linspace(bounds[1], bounds[0], GRID_SIZE)
lon = np.linspace(bounds[2], bounds[3], GRID_SIZE)
if not (source_lat[-1] <= lat[-1] < lat[0] <= source_lat[0]
        and source_lon[0] <= lon[0] < lon[-1] <= source_lon[-1]):
    raise ValueError('The bundled crop does not cover this extent. Refresh the data for the chosen bounds.')

def sample_elevation(latitudes, longitudes):
    rows = (source_lat[0] - np.asarray(latitudes)) / (source_lat[0] - source_lat[1])
    cols = (np.asarray(longitudes) - source_lon[0]) / (source_lon[1] - source_lon[0])
    i = np.clip(np.floor(rows).astype(int), 0, crop.shape[0] - 2)
    j = np.clip(np.floor(cols).astype(int), 0, crop.shape[1] - 2)
    fy, fx = rows - i, cols - j
    return ((1-fy) * ((1-fx)*crop[i,j] + fx*crop[i,j+1])
            + fy * ((1-fx)*crop[i+1,j] + fx*crop[i+1,j+1]))

latitude, longitude = np.meshgrid(lat, lon, indexing='ij')
height_m = np.round(sample_elevation(latitude, longitude), 1)
x_km = np.round((lon - SUMMIT_LON) * km_lon, 5)
y_km = np.round((lat - SUMMIT_LAT) * km_lat, 5)
height_km = height_m / 1000
summit_dem = float(sample_elevation(SUMMIT_LAT, SUMMIT_LON))
peak_row, peak_col = np.unravel_index(np.argmax(height_m), height_m.shape)
peak_height_m = float(height_m[peak_row, peak_col])
peak_latitude = float(latitude[peak_row, peak_col])
peak_longitude = float(longitude[peak_row, peak_col])
peak_x_km = float(x_km[peak_col])
peak_y_km = float(y_km[peak_row])
if height_m.shape != (GRID_SIZE, GRID_SIZE) or not np.isfinite(height_m).all():
    raise ValueError('The display grid is incomplete or invalid.')
if not 1000 < summit_dem < 1900:
    raise ValueError(f'Unexpected elevation at Manam summit coordinate: {summit_dem:.0f} m')
if height_m[peak_row, peak_col] != height_m.max():
    raise ValueError('Summit marker does not point to the highest displayed terrain sample.')
sea_level_km = np.where(height_m <= 0, 0.0, np.nan).astype(np.float32)
print(f'Displayed elevation and bathymetry: {height_m.min():,.0f} to {height_m.max():,.0f} m')
print(f'Sea-level display cells: {np.count_nonzero(height_m <= 0):,} of {height_m.size:,}')
print(f'Displayed DEM peak: {peak_height_m:,.0f} m at {peak_latitude:.5f} S, {peak_longitude:.5f} E')
print(f'DEM at named summit coordinate: {summit_dem:,.0f} m; independent summit reference: {SUMMIT_M:,} m')
```

```python
# Shared scale, including ocean depths; the sea-level surface is a separate trace.
BG, FG, MUTED = '#0d1117', '#eaf0f5', '#a4b2c1'
colors = [[0, '#102b46'], [.16, '#174b6b'], [.32, '#216f89'],
          [.44, '#55a4a4'], [.47, '#4b7554'], [.62, '#89986d'],
          [.78, '#b8a77a'], [.91, '#ded5bb'], [1, '#fff8e8']]
home_camera = dict(eye=dict(x=1.3, y=1.65, z=.85), up=dict(x=0, y=0, z=1),
                   center=dict(x=0, y=0, z=-.12), projection=dict(type='perspective'))
top_camera = dict(eye=dict(x=0, y=0, z=2.5), up=dict(x=0, y=1, z=0),
                  center=dict(x=0, y=0, z=0), projection=dict(type='orthographic'))
fig = go.Figure(go.Surface(
    x=x_km.astype(np.float32), y=y_km.astype(np.float32), z=height_km.astype(np.float32),
    surfacecolor=height_m.astype(np.float32), colorscale=colors, cmin=-1500, cmax=SUMMIT_M,
    customdata=np.stack((np.abs(np.round(latitude, 5)), np.round(longitude, 5)), axis=-1).astype(np.float32),
    hovertemplate='Elevation %{z:.3f} km<br>%{customdata[0]:.5f} S | %{customdata[1]:.5f} E<extra>Terrain</extra>',
    colorbar=dict(title=dict(text='Elevation / depth (m)', side='top'), orientation='h',
                  x=.5, xanchor='center', y=-.08, len=.66, thickness=10,
                  tickvals=[-1500, -1000, -500, 0, 500, 1000, 1500], outlinewidth=0),
    contours=dict(z=dict(show=False, start=-1.5, end=1.8, size=.1,
                         usecolormap=False, color='#45515c', width=1, highlight=False)),
    lighting=dict(ambient=.55, diffuse=.85, roughness=.9, specular=.15, fresnel=.15),
    lightposition=dict(x=10000, y=15000, z=20000), name='Terrain and sea floor'))
fig.add_trace(go.Surface(
    x=x_km.astype(np.float32), y=y_km.astype(np.float32), z=sea_level_km,
    surfacecolor=np.zeros_like(height_m, dtype=np.float32),
    colorscale=[[0, '#287d91'], [1, '#287d91']], cmin=0, cmax=1,
    opacity=.82, showscale=False, connectgaps=False,
    lighting=dict(ambient=1, diffuse=0, specular=0), name='Sea level',
    hovertemplate='Sea level - 0 m<extra>Ocean</extra>'))
fig.add_trace(go.Scatter3d(x=[peak_x_km], y=[peak_y_km], z=[peak_height_m/1000], mode='markers+text',
    marker=dict(size=4, color='#f3d37b', line=dict(width=1, color=BG)),
    text=[f'Manam summit (ref. {SUMMIT_M:,} m)'], textposition='top center', textfont=dict(size=12, color=FG),
    hovertemplate=f'Manam summit<br>Terrain here (SRTM DEM): {peak_height_m:,.0f} m<br>Published summit reference: {SUMMIT_M:,} m<extra>Manam</extra>',
    showlegend=False))
axis = dict(showbackground=False, gridcolor='#28323d', zeroline=False,
            tickfont=dict(size=10, color=MUTED), showspikes=False)
fig.update_layout(template='plotly_dark', paper_bgcolor=BG, plot_bgcolor=BG,
    font=dict(family='Arial, sans-serif', color=FG, size=12), height=700,
    margin=dict(l=10, r=10, t=60, b=80), showlegend=False,
    scene=dict(bgcolor=BG, aspectmode='data', camera=home_camera,
        xaxis=dict(**axis, title='East (km)', range=[-8,8], dtick=4),
        yaxis=dict(**axis, title='North (km)', range=[-8,8], dtick=4),
        zaxis=dict(**axis, title='Altitude (km)', range=[-1.6,2.0], dtick=.5),
    ),
    updatemenus=[dict(type='buttons', direction='right', x=0, xanchor='left', y=1.11,
        yanchor='top', bgcolor='#1b2733', bordercolor='#455569', font=dict(size=12),
        showactive=False, buttons=[
            dict(label='Reset 3D', method='relayout', args=[{'scene.camera':home_camera}]),
            dict(label='Overhead', method='relayout', args=[{'scene.camera':top_camera}]),
            dict(label='Contours', method='restyle', args=[{'contours.z.show':True}, [0]],
                 args2=[{'contours.z.show':False}, [0]])])])
config = dict(responsive=True, scrollZoom=True, displaylogo=False, displayModeBar=False)
fig.show(config=config)
```
