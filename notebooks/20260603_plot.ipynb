```python
import pandas as pd
import plotly.express as px
from pathlib import Path
```

```python
data = pd.DataFrame({
    "year": [2022, 2023, 2024, 2025, 2026],
    "projects": [1, 2, 4, 7, 11],
})

data
```

```python
fig = px.line(
    data,
    x="year",
    y="projects",
    markers=True,
    title="phey.app prototype growth",
)

fig.update_layout(
    template="plotly_dark",
    xaxis_title="Year",
    yaxis_title="Interactive projects",
    width=None,
    height=650,
    autosize=True,
)

fig.show()
```

```python
current = Path.cwd().resolve()
project_root = current
while project_root.name != "p00" and project_root.parent != project_root:
    project_root = project_root.parent
if project_root.name != "p00":
    raise RuntimeError(f"Could not find a parent folder named 'p00' from {current}")

output_dir = project_root / "output"
output_dir.mkdir(exist_ok=True)
output_file = output_dir / "20261003_plot.html"

plot_html = fig.to_html(
    include_plotlyjs=False,
    full_html=False,
    config={
        "responsive": True,
        "displaylogo": False,
    },
)

html = f"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Growth plot | phey.app</title>
  <style>
    html,
    body {{
      margin: 0;
      padding: 0;
      min-height: 100%;
      background: #0d1117;
      color: #f5f7fa;
      font-family: system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", sans-serif;
    }}

    .plot-page {{
      min-height: 100vh;
      background: #0d1117;
      padding: clamp(0.75rem, 2vw, 1.5rem);
      box-sizing: border-box;
    }}

    .plot-wrap {{
      width: 100%;
      margin: 0 auto;
    }}

    a.back-link {{
      display: inline-flex;
      margin-bottom: 1rem;
      color: #f5f7fa;
      text-decoration: none;
      opacity: 0.75;
      font-size: 0.95rem;
    }}

    a.back-link:hover {{
      opacity: 1;
    }}
  </style>
</head>
<body>
  <main class="plot-page">
    <div class="plot-wrap">
      <a class="back-link" href="../index.html">← Back to phey.app</a>
      {plot_html}
    </div>
  </main>
</body>
</html>
"""

# Shared legal navigation and local Plotly dependency survive regeneration.
import sys
if str(project_root / "scripts") not in sys.path:
    sys.path.insert(0, str(project_root / "scripts"))
from site_export import write_plot_page
write_plot_page(output_file, html)

print(f"Exported plot to: {output_file.resolve()}")
```
