```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from pathlib import Path
```

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 801)
data = pd.DataFrame({"x": x, "sin(x)": np.sin(x), "cos(x)": np.cos(x)})

data
```

```python
fig = go.Figure()
for function, color in [("sin(x)", "#58a6ff"), ("cos(x)", "#636efa")]:
    fig.add_trace(go.Scatter(
        x=data["x"], y=data[function], name=function, mode="lines",
        line=dict(color=color, width=3),
        hovertemplate="x: %{x:.2f} rad<br>" + function + ": %{y:.3f}<extra></extra>",
    ))

fig.update_layout(
    template="plotly_dark",
    paper_bgcolor="#0d1117", plot_bgcolor="#0d1117",
    autosize=True, height=480,
    margin=dict(l=55, r=25, t=90, b=60),
    title=dict(text="Sine & cosine", x=0.03, xanchor="left", font=dict(size=20)),
    font=dict(size=13, color="#f5f7fa"),
    legend=dict(orientation="h", x=1, xanchor="right", y=1.12, yanchor="bottom"),
    xaxis=dict(
        title="x (radians)", range=[-2 * np.pi, 2 * np.pi],
        tickmode="array", tickvals=np.arange(-4, 5) * np.pi / 2,
        ticktext=["−2π", "−3π/2", "−π", "−π/2", "0", "π/2", "π", "3π/2", "2π"],
        gridcolor="#21262d", zerolinecolor="#484f58", automargin=True,
    ),
    yaxis=dict(
        title="Amplitude", range=[-1.15, 1.15], dtick=0.5,
        gridcolor="#21262d", zerolinecolor="#484f58", automargin=True,
    ),
)

fig.show()
```
