```python
import pandas as pd
import plotly.express as px
from pathlib import Path
```

```python
data = pd.DataFrame({
    "year": [2022, 2023, 2024, 2025, 2026],
    "projects": [1, 2, 4, 7, 11],
})

data
```

```python
fig = px.line(
    data,
    x="year",
    y="projects",
    markers=True,
    title="phey.app prototype growth",
)

fig.update_layout(
    template="plotly_dark",
    xaxis_title="Year",
    yaxis_title="Interactive projects",
)

fig.show()
```

```python
from pathlib import Path

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)

output_file = output_dir / "20261003_plot.html"

fig.write_html(
    output_file,
    include_plotlyjs="cdn",
    full_html=True,
)

print(f"Exported plot to: {output_file.resolve()}")
```
